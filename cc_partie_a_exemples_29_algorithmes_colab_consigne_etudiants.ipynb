Présentation (Q1)

### Présentation

**Régression linéaire** (*Linear Regression*) : famille **régression**. Elle prédit une valeur
numérique à partir de variables explicatives. Le modèle suppose que la cible est une combinaison
linéaire des variables : ŷ = β₀ + β₁x₁ + … + βₚxₚ.

**Étapes** : données (X, y) en entrée → apprentissage : on cherche les coefficients β qui
minimisent la somme des carrés des erreurs → prédiction ŷ pour de nouvelles données.

**Critère essentiel** : minimiser Σ(yᵢ − ŷᵢ)², où yᵢ est la valeur observée, ŷᵢ la valeur
prédite, β₀ l'ordonnée à l'origine et βⱼ le coefficient de la variable xⱼ.
Les β sont les **paramètres** (appris). Cette version simple n'a **pas d'hyperparamètre** à régler.

**Avantages** : simple et rapide ; interprétable (chaque β donne l'effet d'une variable).
**Limites** : suppose une relation linéaire ; sensible aux valeurs extrêmes et aux variables corrélées entre elles.

```python
import numpy as np, matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
np.random.seed(42)
X_demo = np.linspace(0, 10, 30).reshape(-1, 1)
y_demo = 2.5 * X_demo.ravel() + 5 + np.random.normal(0, 3, 30)
reg = LinearRegression().fit(X_demo, y_demo)
plt.figure(figsize=(7, 4))
plt.scatter(X_demo, y_demo, color='blue', label='Données réelles')
plt.plot(X_demo, reg.predict(X_demo), color='red', linewidth=2, label='Droite ajustée')
plt.title("Régression linéaire : Minimisation de la somme des carrés des erreurs")
plt.xlabel("X"); plt.ylabel("y"); plt.legend(); plt.grid(True); plt.show()
```

Le nuage bleu
représente les observations réelles. La ligne rouge montre la droite ajustée qui minimise la somme des distances verticales au carré (lignes pointillées grises représentant les résidus $y_i - \hat{y}_i$).

Contextes d'utilisationCas d'usage 1 : Finance / Gestion — Estimation de la structure des coûts fixes et variables d'une entreprise en fonction du volume de production.   Cas d'usage 2 : Marketing — Prédiction des ventes trimestrielles en fonction des dépenses d'acquisition d'utilisateurs.   Cas inapproprié : Prédiction de séries temporelles financières complexes présentant une volatilité extrême ou des régimes non linéaires.   Métriques adaptées : Erreur absolue moyenne (MAE), Erreur quadratique moyenne (RMSE), Coefficient de détermination ($R^2$).   Analyse de l'expérience PythonEntrées et hyperparamètres : Modèle ajusté par la méthode OLS sur les variables Xr_ent et yr_ent sans hyperparamètres.   Analyse du résultat exécuté : La cellule calcule la MAE sur l'échantillon de test Xr_test. L'évaluation de l'erreur moyenne reflète la variance irréductible ajoutée synthétiquement ($\sigma=13$).   Limites scientifiques de la sortie : La MAE indique l'écart moyen mais ne permet pas de vérifier les hypothèses de Gauss-Markov (homoscédasticité, normalité des résidus), ni d'exclure un biais de sous-apprentissage si la relation réelle n'est pas linéaire.

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

PrésentationNom : Régression Ridge / Ridge Regression (Régularisation $L_2$).   Famille : Apprentissage supervisé — Régression.   Intuition : Ajouter une pénalité quadratique sur l'amplitude des coefficients de la régression linéaire pour éviter le surapprentissage et réduire l'impact de la multicolinéarité.   Formule essentielle :$$\min_{\beta} \left( \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \alpha \sum_{j=1}^{p} \beta_j^2 \right)$$
Paramètres : $\beta_j$. Hyperparamètre : $\alpha \ge 0$ (intensité de la pénalité $L_2$).   Avantages : Stabilise l'estimation en présence de variables explicatives très corrélées ; réduit la variance des prédictions.   Limites : Conserve l'intégralité des variables (ne réalise pas de sélection explicite) ; exige une normalisation préalable des données.  

```python
import numpy as np, matplotlib.pyplot as plt
from sklearn.linear_model import Ridge
np.random.seed(42)
X_demo = np.linspace(-3, 3, 20).reshape(-1, 1)
y_demo = 2 * X_demo.ravel() + np.random.normal(0, 2, 20)
alphas = [0, 10, 100]
plt.figure(figsize=(7, 4))
plt.scatter(X_demo, y_demo, color='black', label='Points')
for a in alphas:
    r = Ridge(alpha=a).fit(X_demo, y_demo)
    plt.plot(X_demo, r.predict(X_demo), label=f'alpha={a}')
plt.title("Impact du paramètre de régularisation L2 (Ridge)")
plt.xlabel("X"); plt.ylabel("y"); plt.legend(); plt.grid(True); plt.show()
```

Lecture du schéma : À mesure que l'hyperparamètre $\alpha$ augmente, la pente de la droite diminue vers zéro, ce qui réduit la sensibilité du modèle aux fluctuations individuelles de l'échantillon.Contextes d'utilisationCas d'usage 1 : Gestion du risque / Finance — Modélisation du score de crédit incluant de nombreux ratios financiers très corrélés.   Cas d'usage 2 : Econométrie — Prévision des indicateurs économiques où le nombre de variables dépasse le nombre d'observations ($p > n$).   Cas inapproprié : Problèmes où une réduction stricte du nombre de variables est nécessaire pour l'explicabilité métier.   Métriques adaptées : MAE, RMSE, $R^2$.   Analyse de l'expérience PythonEntrées et hyperparamètres : Pipeline composé de StandardScaler() et Ridge(alpha=1.0).   Analyse du résultat exécuté : La MAE calculée sur Xr_test évalue la performance après standardisation des entrées et régularisation $L_2$. La normalisation garantit un traitement uniforme de la pénalité.   Limites scientifiques de la sortie : Le résultat affiche la MAE pour une seule valeur $\alpha=1.0$. Il ne présente pas le chemin de régularisation ni une validation croisée (RidgeCV) permettant de valider l'optimum de l'hyperparamètre.

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

PrésentationNom : Lasso / Least Absolute Shrinkage and Selection Operator (Régularisation $L_1$).   Famille : Apprentissage supervisé — Régression.   Intuition : Pénaliser la régression par la somme des valeurs absolues des coefficients, contraignant certains coefficients à s'annuler exactement afin de sélectionner automatiquement les variables les plus pertinentes.   Formule essentielle :$$\min_{\beta} \left( \frac{1}{2n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \alpha \sum_{j=1}^{p} \vert{}\beta_j\vert{} \right)$$
Paramètres : $\beta_j$. Hyperparamètres : $\alpha$ (poids de la pénalité $L_1$), max_iter.   Avantages : Sélection automatique de variables ; génère des modèles creux (sparse) faciles à interpréter.   Limites : Choisit arbitrairement une seule variable au sein d'un groupe de variables fortement corrélées ; instable si $p >> n$.  

```python
import numpy as np, matplotlib.pyplot as plt
from sklearn.linear_model import Lasso
np.random.seed(42)
X_demo = np.random.randn(50, 5)
y_demo = 3 * X_demo[:, 0] - 2 * X_demo[:, 1] + np.random.randn(50)
model = Lasso(alpha=0.3).fit(X_demo, y_demo)
plt.figure(figsize=(7, 4))
plt.bar(range(5), model.coef_, color='teal')
plt.axhline(0, color='black', linewidth=1)
plt.title("Sélection de variables par Lasso (Coefficients annulés)")
plt.xlabel("Index des variables"); plt.ylabel("Valeur des coefficients"); plt.grid(True); plt.show()
```

Contextes d'utilisationCas d'usage 1 : Commerce international — Sélection des déterminants majeurs du volume d'exportation parmi un grand ensemble d'indicateurs macroéconomiques.   Cas d'usage 2 : Marketing — Identification des canaux d'acquisition digitaux les plus efficaces parmi plusieurs dizaines de leviers.   Cas inapproprié : Données à très forte colinéarité où toutes les variables doivent conserver une contribution positive.   Métriques adaptées : $R^2$, MAE, RMSE.   Analyse de l'expérience PythonEntrées et hyperparamètres : Pipeline avec StandardScaler() et Lasso(alpha=0.1, max_iter=10000).   Analyse du résultat exécuté : La cellule retourne le coefficient de détermination $R^2$ mesuré sur Xr_test, indiquant la part de variance expliquée.   Limites scientifiques de la sortie : Bien que le score $R^2$ valide la capacité prédictive globale, la sortie n'affiche pas les coefficients $\beta_j$, empêchant de vérifier quelles variables ont été réellement sélectionnées ou éliminées par le Lasso.

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),PolynomialFeatures(degree=2,include_bias=False),Ridge(alpha=10)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

PrésentationNom : Régression polynomiale / Polynomial Regression.   Famille : Apprentissage supervisé — Régression.   Intuition : Étendre l'espace des descripteurs en générant leurs puissances et leurs interactions pour modéliser des relations non linéaires à l'aide d'un modèle linéaire.   Formule essentielle :$$y = \beta_0 + \beta_1 x + \beta_2 x^2 + \dots + \beta_d x^d + \epsilon$$
Paramètres : coefficients $\beta_j$. Hyperparamètres : degree ($d$), paramètre de régularisation.   Avantages : Modélise efficacement des courbures sans changer le cadre de la régression linéaire.   Limites : Croissance rapide de la dimension des caractéristiques ; risque élevé de surapprentissage aux limites du domaine.

```python
import numpy as np, matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline

X_demo = np.linspace(-3, 3, 30).reshape(-1, 1)
y_demo = X_demo.ravel()**2 + np.random.normal(0, 1, 30)
plt.figure(figsize=(7, 4))
plt.scatter(X_demo, y_demo, color='darkorange', label='Données')
for d in [1, 2, 5]:
    m = make_pipeline(PolynomialFeatures(degree=d), LinearRegression()).fit(X_demo, y_demo)
    plt.plot(X_demo, m.predict(X_demo), label=f'Degré {d}')
plt.title("Ajustement par Régression Polynomiale")
plt.xlabel("X"); plt.ylabel("y"); plt.legend(); plt.grid(True); plt.show()
```

Lecture du schéma : Un degré 1 (droite) sous-ajuste la parabole. Le degré 2 s'ajuste correctement à la courbure, tandis qu'un degré élevé (5) oscille et risque d'ajuster le bruit.Contextes d'utilisationCas d'usage 1 : Logistique — Estimation des coûts de transport non linéaires présentant un effet d'échelle ou de saturation.   Cas d'usage 2 : Gestion de la production — Modélisation du rendement d'un processus industriel en fonction de la température.   Cas inapproprié : Données à grande dimension initiale ($p > 50$) où le nombre de termes croisés d'ordre supérieur devient ingérable.   Métriques adaptées : MAE, RMSE, $R^2$.   Analyse de l'expérience PythonEntrées et hyperparamètres : Pipeline avec StandardScaler(), PolynomialFeatures(degree=2, include_bias=False) et Ridge(alpha=10).   Analyse du résultat exécuté : La MAE mesure la précision de la prédiction du modèle quadratique régularisé sur le jeu de test.   Limites scientifiques de la sortie : Le résultat affiche une MAE unique sans la comparer à celle de la régression linéaire simple. Il n'est donc pas possible de valider formellement l'apport théorique du degré 2 par rapport à un modèle linéaire sans terme croisé.   

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.tree import DecisionTreeRegressor
modele=DecisionTreeRegressor(max_depth=5,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

résentationNom : Arbre de décision pour la régression / Decision Tree Regressor.   Famille : Apprentissage supervisé — Régression.   Intuition : Partitionner récursivement l'espace des caractéristiques en sous-régions orthogonales en choisissant les coupures qui minimisent la variance intra-région.   Formule essentielle :$$\text{Critère de séparation (MSE)} : \text{Var}(S) = \frac{1}{\vert{}S\vert{}} \sum_{i \in S} (y_i - \bar{y}_S)^2$$
Paramètres : Seuils de coupure et structure de l'arbre. Hyperparamètres : max_depth, min_samples_split, min_samples_leaf.   Avantages : Structure très lisible ; capture naturellement les non-linéarités et les interactions sans prétraitement complexe.   Limites : Forte instabilité (variance élevée) ; produit des prédictions constantes par morceaux incapables d'extrapoler.  

```python
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor, plot_tree
import numpy as np
X_demo = np.linspace(0, 10, 20).reshape(-1, 1)
y_demo = np.sin(X_demo.ravel()) + np.random.normal(0, 0.2, 20)
tree = DecisionTreeRegressor(max_depth=2, random_state=42).fit(X_demo, y_demo)
plt.figure(figsize=(8, 4))
plot_tree(tree, feature_names=['X'], filled=True, rounded=True)
plt.title("Structure d'un Arbre de Régression (Profondeur = 2)")
plt.show()
```

Lecture du schéma : Chaque nœud effectue un test de type $X \le \text{seuil}$. Les feuilles terminales contiennent la valeur prédictive moyenne des individus associés.   Contextes d'utilisationCas d'usage 1 : Gestion / Logistique — Estimation des besoins en stocks en fonction de variables catégorielles et temporelles.   Cas d'usage 2 : Immobilier — Évaluation de la valeur locative à partir de règles conditionnelles claires.   Cas inapproprié : Estimation de tendances linéaires continues strictes nécessitant des interpolations ou extrapolations lisses.   Métriques adaptées : MAE, RMSE, $R^2$.   Analyse de l'expérience PythonEntrées et hyperparamètres : DecisionTreeRegressor(max_depth=5, random_state=42) ajusté sur Xr_ent.   Analyse du résultat exécuté : La cellule calcule la MAE sur Xr_test. La contrainte de profondeur max_depth=5 limite l'expansion de l'arbre pour contrôler la variance.   Limites scientifiques de la sortie : La valeur de MAE affichée ne permet pas de déterminer si la profondeur 5 est sous-optimale (sous-apprentissage) ou si l'arbre présente une sensibilité trop forte aux données d'entraînement.

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.ensemble import RandomForestRegressor
modele=RandomForestRegressor(n_estimators=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

PrésentationNom : Forêt aléatoire pour la régression / Random Forest Regressor.   Famille : Apprentissage supervisé — Régression (Ensemble/Bagging).   Intuition : Entraîner plusieurs arbres de régression indépendants sur des sous-échantillons bootstrap et faire la moyenne de leurs prédictions pour réduire la variance.   Formule essentielle :$$\hat{y}_{\text{RF}} = \frac{1}{B} \sum_{b=1}^{B} T_b(x)$$
Paramètres : Ensemble des $B$ arbres de décision. Hyperparamètres : n_estimators, max_depth, max_features, random_state.   Avantages : Forte capacité de généralisation ; très bonne résistance au surapprentissage grâce au bagging.   Limites : Perte d'interprétabilité visuelle directe par rapport à un arbre unique ; temps de calcul et empreinte mémoire plus élevés.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 3))

# Boîtes de texte avec l'argument bbox corrigé
ax.text(0.5, 0.8, "Données d'entraînement",
        bbox=dict(boxstyle="round,pad=0.3", facecolor="lightgray", edgecolor="black"),
        ha="center", va="center")

ax.annotate('', xy=(0.2, 0.5), xytext=(0.5, 0.75), arrowprops=dict(arrowstyle="->", lw=1.5))
ax.annotate('', xy=(0.8, 0.5), xytext=(0.5, 0.75), arrowprops=dict(arrowstyle="->", lw=1.5))

ax.text(0.2, 0.4, "Arbre 1 (Bootstrap 1)",
        bbox=dict(boxstyle="square,pad=0.4", facecolor="lightblue", edgecolor="black"),
        ha="center", va="center")
ax.text(0.8, 0.4, "Arbre B (Bootstrap B)",
        bbox=dict(boxstyle="square,pad=0.4", facecolor="lightblue", edgecolor="black"),
        ha="center", va="center")

ax.annotate('', xy=(0.5, 0.18), xytext=(0.2, 0.35), arrowprops=dict(arrowstyle="->", lw=1.5))
ax.annotate('', xy=(0.5, 0.18), xytext=(0.8, 0.35), arrowprops=dict(arrowstyle="->", lw=1.5))

ax.text(0.5, 0.1, "Prédiction finale = Moyenne des prédictions",
        bbox=dict(boxstyle="round,pad=0.3", facecolor="lightgreen", edgecolor="black"),
        ha="center", va="center")

ax.axis('off')
plt.title("Principe d'agrégation d'une Forêt Aléatoire (Bagging)", fontweight='bold', pad=15)
plt.tight_layout()
plt.show()
```

Lecture du schéma : Chaque arbre génère une prédiction indépendante à partir d'un sous-échantillon bootstrap, puis l'agrégation par la moyenne établit l'estimation finale.Contextes d'utilisationCas d'usage 1 : Finance — Estimation de la valeur résiduelle d'actifs ou de véhicules en fin de contrat de leasing.   Cas d'usage 2 : Assurance — Prédiction du coût moyen des sinistres à partir des caractéristiques des assurés. Analyse de l'expérience PythonEntrées et hyperparamètres : RandomForestRegressor(n_estimators=80, random_state=42).   Analyse du résultat exécuté : La cellule renvoie la MAE calculée sur Xr_test en moyennant les résultats de 80 arbres.   Limites scientifiques de la sortie : Le résultat ne présente ni l'erreur Out-Of-Bag (OOB), ni le graphique d'importance relative des variables (Feature Importances), ce qui limite l'interprétabilité des facteurs explicatifs.

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.ensemble import HistGradientBoostingRegressor
modele=HistGradientBoostingRegressor(max_iter=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

PrésentationNom : HistGradientBoosting / Gradient Boosting Regressor.   Famille : Apprentissage supervisé — Régression (Ensemble/Boosting).   Intuition : Construire séquentiellement des arbres de décision courts où chaque nouvel arbre est ajusté pour corriger les erreurs (résidus) commises par les arbres précédents.   Formule essentielle :$$F_m(x) = F_{m-1}(x) + \nu \cdot h_m(x)$$
Paramètres : Ensemble des arbres séquentiels. Hyperparamètres : learning_rate ($\nu$), max_iter, max_leaf_nodes.   Avantages : Performances prédictives très élevées sur données tabulaires ; vitesse d'apprentissage optimisée par histogrammes.   Limites : Sensible au surapprentissage si le taux d'apprentissage ou le nombre d'itérations est trop élevé ; réglage d'hyperparamètres nécessaire.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 3))

# Textes avec style bbox corrigé
ax.text(0.08, 0.5, "Modèle F_0",
        bbox=dict(boxstyle="square,pad=0.4", facecolor="lightgray", edgecolor="black"),
        ha="center", va="center")

ax.annotate('', xy=(0.24, 0.5), xytext=(0.15, 0.5), arrowprops=dict(arrowstyle="->", lw=1.5))

ax.text(0.36, 0.5, "Calcul des\nRésidus",
        bbox=dict(boxstyle="circle,pad=0.4", facecolor="pink", edgecolor="black"),
        ha="center", va="center")

ax.annotate('', xy=(0.50, 0.5), xytext=(0.45, 0.5), arrowprops=dict(arrowstyle="->", lw=1.5))

ax.text(0.66, 0.5, "Arbre h_1\n(Ajuste résidus)",
        bbox=dict(boxstyle="square,pad=0.4", facecolor="lightblue", edgecolor="black"),
        ha="center", va="center")

ax.annotate('', xy=(0.79, 0.5), xytext=(0.76, 0.5), arrowprops=dict(arrowstyle="->", lw=1.5))

ax.text(0.90, 0.5, "$F_1 = F_0 + \\nu \\cdot h_1$",
        bbox=dict(boxstyle="square,pad=0.4", facecolor="lightgreen", edgecolor="black"),
        ha="center", va="center")

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis('off')
plt.title("Fonctionnement séquentiel du Gradient Boosting", fontweight='bold', pad=15)
plt.tight_layout()
plt.show()
```

ecture du schéma : Chaque arbre supplémentaire $h_m$ est entraîné sur le gradient négatif de la perte afin de réduire séquentiellement les erreurs résiduelles.   Contextes d'utilisationCas d'usage 1 : Marketing — Prédiction de la valeur vie client (Customer Lifetime Value).   Cas d'usage 2 : Logistique — Prévision de la demande de fret sur des hubs de transport.   Cas inapproprié : Petits jeux de données où des modèles simples régularisés offrent une meilleure stabilité sans surapprentissage.   Métriques adaptées : MAE, RMSE, $R^2$.   Analyse de l'expérience PythonEntrées et hyperparamètres : HistGradientBoostingRegressor(max_iter=80, random_state=42).   Analyse du résultat exécuté : La MAE mesure la performance après 80 itérations d'histogrammes sur l'échantillon de test Xr_test.   Limites scientifiques de la sortie : Le résultat ne présente pas la courbe de perte (loss history), empêchant de vérifier l'état de convergence ou de détecter un éventuel surapprentissage.

Lecture du schéma : La fonction sigmoïde transforme la combinaison linéaire $z \in ]-\infty, +\infty[$ en une probabilité valide $P \in ]0, 1[$, étiquetée à 1 si $P \ge 0.5$.   Contextes d'utilisationCas d'usage 1 : Finance — Décision d'octroi de crédit (Accordé / Refusé) selon les caractéristiques du demandeur. ## 13. SVM

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.svm import SVC
modele=make_pipeline(StandardScaler(),SVC(kernel='rbf',C=1.0,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

PrésentationNom : Machine à vecteurs de support / Support Vector Machine (SVC).   Famille : Apprentissage supervisé — Classification.   Intuition : Chercher l'hyperplan qui sépare les classes en maximisant la marge entre cet hyperplan et les points les plus proches (vecteurs de support).   Formule essentielle :$$\min_{w, b} \frac{1}{2} ||w||^2 + C \sum_{i=1}^{n} \xi_i$$
Paramètres : Poids $w$, biais $b$, vecteurs de support. Hyperparamètres : C, kernel ('rbf', 'linear'), gamma.   Avantages : Très performant en grande dimension ; l'astuce du noyau (kernel trick) permet de modéliser des frontières non linéaires complexes. Intuition : Projeter les données sur de nouveaux axes orthogonaux (composantes principales) définis pour conserver le maximum de variance du jeu de données.   Formule essentielle :$$\Sigma = \frac{1}{n} X^T X \quad ; \quad \Sigma v_k = \lambda_k v_k$$
Paramètres : Vecteurs propres $v_k$ et valeurs propres $\lambda_k$. Hyperparamètres : n_components.   Avantages : Réduit la dimensionnalité en éliminant les redondances linéaires ; facilite la visualisation en 2D/3D.   Limites : Les composantes principales sont des combinaisons linéaires parfois complexes à interpréter métier ; ne capture pas les structures non linéaires.

```python
import numpy as np, matplotlib.pyplot as plt
from sklearn.decomposition import PCA
np.random.seed(42)
X_pca = np.dot(np.random.randn(100, 2), [[2, 1], [1, 2]])
pca = PCA(n_components=2).fit(X_pca)
plt.figure(figsize=(7, 4))
plt.scatter(X_pca[:,0], X_pca[:,1], alpha=0.5, label='Données')
for length, vector in zip(pca.explained_variance_, pca.components_):
    v = vector * 2 * np.sqrt(length)
    plt.plot([pca.mean_[0], pca.mean_[0] + v[0]], [pca.mean_[1], pca.mean_[1] + v[1]], 'r-', lw=3)
plt.title("Projection sur les axes principaux de l'ACP")
plt.xlabel("X1"); plt.ylabel("X2"); plt.grid(True); plt.show()
```

Lecture du schéma : Les lignes rouges matérialisent les directions orthogonales de variance maximale le long desquelles les observations sont projetées.Contextes d'utilisationCas d'usage 1 : Finance — Construction d'indices synthétiques de santé financière à partir de nombreux ratios boursiers.   Cas d'usage 2 : Traitement d'images — Compression d'images et réduction du bruit.   Cas inapproprié : Jeux de données comportant des structures non linéaires complexes (ex. manifolds courbés). résentationNom : Multi-Armed Bandits / Manchots à $K$ bras (Estimation naïve).   Famille : Apprentissage par renforcement. Paramètres : Estimations empiriques $\hat{Q}(a)$ et compteurs $N(a)$. Hyperparamètres : Horizon $T$.   Avantages : Cadre théorique adapté aux problèmes de décisions séquentielles simples sans état complexe.   Limites : L'exploration purement alternée gaspille des opportunités en continuant d'évaluer les bras défavorables au même rythme.

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3);epsilon=.1
for t in range(1000):
    bras=int(generateur.integers(3)) if generateur.random()<epsilon else int(np.argmax((somme+1)/(comptes+1)))
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Nombre de choix:',comptes,'Valeurs estimées:',somme/np.maximum(comptes,1))
```

PrésentationNom : Stratégie $\epsilon$-Greedy / Epsilon-Greedy Bandit.   Famille : Apprentissage par renforcement.   Intuition : Équilibrer l'exploration et l'exploitation en choisissant une action au hasard avec la probabilité $\epsilon$, et l'action optimale estimée avec la probabilité $1-\epsilon$.   Formule essentielle :$$A_t = \begin{cases} \text{action aléatoire} & \text{avec probabilité } \epsilon \\ \arg\max_{a} \hat{Q}_t(a) & \text{avec probabilité } 1 - \epsilon \end{cases}$$
Paramètres : Estimations $\hat{Q}(a)$. Hyperparamètres : $\epsilon$ (taux d'exploration).   Avantages : Arbitrage simple et efficace entre exploration et exploitation.   Limites : Lors de l'exploration, l'algorithme teste de manière égale les pires actions et les secondes meilleures.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

résentationNom : SARSA / State-Action-Reward-State-Action (Apprentissage sur-politique / On-Policy TD Control).   Famille : Apprentissage par renforcement.   Intuition : Mettre à jour la valeur $Q(s, a)$ en tenant compte de la prochaine action $a'$ réellement sélectionnée par la politique courante de l'agent. Paramètres : Table $Q(s, a)$. Hyperparamètres : $\alpha$, $\gamma$, $\epsilon$.   Avantages : Prend en compte les risques des actions d'exploration pendant l'apprentissage (politique plus prudente).   Limites : Peut être légèrement sous-optimale par rapport à Q-Learning une fois l'exploration terminée

## Annexes — implémentations manuelles des trois algorithmes imposés

```python
# Régression linéaire par descente de gradient : données univariées
x=np.linspace(-1,1,100); y=2+3*x
poids=0.;biais=0.;pas=.1
for iteration in range(500):
    prediction=poids*x+biais
    erreur=prediction-y
    poids-=pas*2*np.mean(erreur*x)
    biais-=pas*2*np.mean(erreur)
print('Paramètres appris manuellement:',poids,biais)
from sklearn.linear_model import LinearRegression
print('Référence sklearn:',LinearRegression().fit(x.reshape(-1,1),y).coef_,LinearRegression().fit(x.reshape(-1,1),y).intercept_)
```

```python
# K-means manuel : assignation et recalcul jusqu'à stabilité
valeurs=X_clu.copy(); centroids=valeurs[np.random.default_rng(GRAINE).choice(len(valeurs),size=3,replace=False)].copy()
for iteration in range(100):
    distances=((valeurs[:,None,:]-centroids[None,:,:])**2).sum(axis=2)
    groupes=distances.argmin(axis=1)
    nouveaux=np.array([valeurs[groupes==k].mean(axis=0) if (groupes==k).any() else centroids[k] for k in range(3)])
    if np.allclose(nouveaux,centroids):break
    centroids=nouveaux
print('Centroïdes manuels:',centroids)
from sklearn.cluster import KMeans
print('Centroïdes sklearn (ordre arbitraire):',KMeans(n_clusters=3,n_init=10,random_state=GRAINE).fit(valeurs).cluster_centers_)
```
